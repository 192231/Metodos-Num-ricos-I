#**Iteracion de punto fijo**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return 3*x**2-np.exp(x)
```

```python
xx=np.linspace(-1,1)
plt.plot(xx,f(xx))
plt.axhline(y=0, color='pink')
plt.axvline(x=0, color='pink')
```

```python
def g(x):
    return np.sqrt(np.exp(x)/3)
```

```python
g(0)
```

```python
def pfijo(p0,i):

   p0=0.5
   lista=[]
   error=[]

   for i in range(10):
      p1=g(p0)
      lista.append(p1)
      e=np.abs(p1-p0)
      error.append(e)
      p0=p1
   return lista, error
```

```python
p=pfijo(0.5,15)
```
